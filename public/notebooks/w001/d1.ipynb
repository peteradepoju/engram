```python
import numpy as np

# Simulate a leaky integrate-and-fire neuron for 300 ms.
tau, R = 20.0, 10.0                     # ms, megaohms
v_rest, v_th, v_reset = -70.0, -54.0, -75.0
dt, T = 0.1, 300.0                      # ms
I = 2.2                                 # nA, switched on from 50 to 250 ms

v = v_rest
spikes = []
for k in range(int(T / dt)):
    t = k * dt
    current = I if 50 <= t <= 250 else 0.0
    v += dt / tau * (-(v - v_rest) + R * current)
    if v >= v_th:
        spikes.append(t)
        v = v_reset

print(f"{len(spikes)} spikes")
print("first spike times (ms):", np.round(spikes[:5], 1))
```
