```python
import numpy as np
import matplotlib.pyplot as plt

def simulate(w, I_A=2.5, T=300.0, dt=0.1):
    tau, R = 20.0, 10.0
    v_rest, v_th, v_reset = -70.0, -54.0, -75.0
    vA = vB = v_rest
    spikes_A, spikes_B, trace_B = [], [], []
    for k in range(int(T / dt)):
        t = k * dt
        vA += dt / tau * (-(vA - v_rest) + R * I_A)
        vB += dt / tau * (-(vB - v_rest))
        if vA >= v_th:
            spikes_A.append(t); vA = v_reset
            vB += w                      # the synapse: a jump in B's voltage
        if vB >= v_th:
            spikes_B.append(t); vB = v_reset
        trace_B.append(vB)
    return spikes_A, spikes_B, np.array(trace_B)

sA, sB, trace = simulate(w=6.0)
print(f"A fired {len(sA)} times, B fired {len(sB)} times")
plt.figure(figsize=(6, 2.5))
plt.plot(np.arange(len(trace)) * 0.1, trace)
plt.xlabel("time (ms)"); plt.ylabel("B voltage (mV)")
plt.show()
```
