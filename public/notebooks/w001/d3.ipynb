```python
import numpy as np
def test_function():
    "summary returns three numbers"
    out = summary(np.array([1.0, 2.0, 3.0, 4.0]))
    assert len(out) == 3

def test_values():
    "mean, SD (n-1) and SEM are correct"
    m, s, e = summary(np.array([2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0]))
    assert abs(m - 5.0) < 1e-9, "Check the mean."
    assert abs(s - np.std([2,4,4,4,5,5,7,9], ddof=1)) < 1e-9, "Use n - 1 in the SD (ddof=1 in NumPy)."
    assert abs(e - s / np.sqrt(8)) < 1e-9, "SEM is SD divided by the square root of n."

test_function()
test_values()
print("All checks passed")
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(7, 3), sharey=True)
for ax, kind in zip(axes, ["SD", "SEM"]):
    for i, (name, x) in enumerate([("face", face), ("house", house)]):
        m, s, e = summary(x)
        ax.bar(i, m, color=["#c2416b", "#1f6fc7"][i], alpha=0.8)
        ax.errorbar(i, m, yerr=s if kind == "SD" else e, color="k", capsize=6)
        ax.plot(np.full(len(x), i) + rng.uniform(-0.15, 0.15, len(x)), x, ".", color="0.3", ms=4)
    ax.set_xticks([0, 1], ["face", "house"]); ax.set_title(f"error bars: {kind}")
axes[0].set_ylabel("firing rate (Hz)")
plt.tight_layout(); plt.show()
```
