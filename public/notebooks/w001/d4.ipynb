### Exercise: Bug hunt

This function should scale values to between 0 and 1 without touching its input. It has two bugs. Find and fix them.

```python
def normalise(values):
    lo, hi = min(values), max(values)
    for i in range(len(values)):
        values[i] = (values[i] - lo) / hi
    return values

print(normalise([2.0, 4.0, 6.0]))
```

```python
def test_fixed():
    "normalise scales values to the range 0 to 1"
    out = normalise([2.0, 4.0, 6.0])
    assert out == [0.0, 0.5, 1.0], f"Got {out}"

def test_unchanged_input():
    "the input list is not changed"
    data = [3.0, 1.0, 2.0]
    normalise(data)
    assert data == [3.0, 1.0, 2.0], "normalise changed the list it was given. Never change raw data in place."

test_fixed()
test_unchanged_input()
print("All checks passed")
```
