```python
import numpy as np

def fires(I1, I2, T=200.0, dt=0.1):
    tau, R, v_rest, v_th, v_reset = 20.0, 10.0, -70.0, -54.0, -75.0
    I = 0.8 * I1 + 1.2 * I2
    v = v_rest
    for _ in range(int(T / dt)):
        v += dt / tau * (-(v - v_rest) + R * I)
        if v >= v_th:
            return 1
    return 0

rng = np.random.default_rng(3)
inputs = rng.uniform(0, 2.5, size=(200, 2))
labels = np.array([fires(a, b) for a, b in inputs])
print("fraction of trials with a spike:", labels.mean())
```
